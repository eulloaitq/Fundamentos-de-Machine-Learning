## Evaluación de clasificación

**Objetivo:** Analizar una variable cualitativa ordinal denominada **nivel de estudios** que contiene 500 registros mediante frecuencias y gráficos. Para luego elaborar un análisis similar utilizando la biblioteca pandas.

## Importaciones y semilla

Se importan las librerias:
- **numpy** Esta ofrece operaciones sobre arreglos y permite permutar los datos.
- **random** Es el módulo de aleatoriedad de Python.
- **random.seed(42)** fija la semilla del módulo **random**.

```python
# Creación aleatoria de datos
import numpy as np
import random
random.seed(42)
```

## Construcción de 500 observaciones

La variable **nivel_estudios** se define como una lista con cero elementos, luego la función **extend** incorpora tantas repeticiones de cada categoría como indica el factor: 110 de Secundaria, 172 de Bachillerato, 145 de Grado, 52 de Master y 21 de Doctorado, obteniendo en total 500 observaciones, finalmente **np.random.permutation** reorganiza el orden de los datos sin cambiar las cantidades.

```python
nivel_estudios = []
nivel_estudios.extend(['Secundaria']*110)
nivel_estudios.extend(['Bachillerato']*172)
nivel_estudios.extend(['Grado']*145)
nivel_estudios.extend(['Master']*52)
nivel_estudios.extend(['Doctorado']*21)
nivel_estudios = np.random.permutation(nivel_estudios)
```

## Visualizaciòn de los datos

Se muestran los 500 valores de la variable.

```python
print(nivel_estudios)
```

## Frecuencia absoluta con NumPy

La **np.unique(..., return_counts=True)** obtiene dos arreglos: el uno de **estudios**, que contiene las distintas catergorias y el otro con la **frecuencia**, que contiene el número de veces que aparece cada categoría.

```python
# DIAGRAMA DE BARRAS
import matplotlib.pyplot as plt
import numpy as np
estudios, frecuencia = np.unique(nivel_estudios, return_counts=True)
print(estudios, frecuencia)
```

## Gráfico de barras

El comando **plt.ylabel** define la etiqueta del eje vertical con la frecuencia absoluta, **plt.xlabel** etiqueta al eje horizontal con el nivel de estudios, **plt.title** agrega el título al gráfico y **plt.bar** dibuja una barra por cada nivel educativo, **plt.show()** presenta el gráfico.

```python
plt.ylabel('Frecuencia absoluta')
plt.xlabel('Nivel de estudios')
plt.title('Diagrama de barras')
plt.bar(estudios, frecuencia) # https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.bar.html
plt.show()
```

## Gráfico de sectores

El gráfico circular o de pastel permite apreciar qué parte del total corresponde a cada nivel, **labels** agrega nombres a cada partición y **autopct='%1.2f%%'** muestra los porcentajes con formato de dos decimales.

```python
# DIAGRAMA DE SECTORES
colores = {'red', 'blue', 'orange', 'green', 'cyan'}
plt.pie(frecuencia, labels=estudios, colors=colores, autopct='%1.2f%%')
plt.title('Diagrama de sectores')
plt.show()
```

# Tarea implementar con pandas esta información
## Implementación con Pandas

## Importar Pandas y crear un DataFrame

Se ejecutan los siguientes comandos:
- **import pandas as pd:** Importa la biblioteca con su alias más común.
- **pd.DataFrame(...):** Organiza los registros en una tabla con un atributo denominado **Nivel de estudios**.
- **head(10):** Muestra los primeros diez registros para verificar la estructura.

Un DataFrame es una matriz que contiene las instancias y los atributos para el análisis de los datos.

```python
import pandas as pd

df = pd.DataFrame({'Nivel de estudios': nivel_estudios})
print('Primeros 10 registros:')
print(df.head(10))
```

## Comprobación del número de registros

El comando **len(df)** devuelve la cantidad de instancis del DataFrame, aquí se confirma que la tabla conserva los 500 datos originales.

```python
print('Total de registros:', len(df))
```

## Calcular frecuencias absolutas con Pandas

El comando **value_counts()** muestra el valor de cuántas veces aparece cada categoría, **sort_index()** ordena las etiquetas alfabéticamente, facilitando la comparación con la salida original de NumPy.

```python
frecuencia_pd = df['Nivel de estudios'].value_counts().sort_index()
print('Frecuencia absoluta por nivel de estudios:')
print(frecuencia_pd)
```

## Frecuencias relativas y porcentajes

EL comando **value_counts(normalize=True)** calcula la proporción de cada categoría respecto del total, al multiplicar ese resultado por 100 se transforma esa proporción en porcentaje.

```python
frecuencia_relativa_pd = df['Nivel de estudios'].value_counts(normalize=True).sort_index()
porcentaje_pd = frecuencia_relativa_pd * 100
print('Frecuencia relativa:')
print(frecuencia_relativa_pd)
print('Porcentaje por categoría:')
print(porcentaje_pd.round(2))
```

## Tabla de frecuencias completa

Se crea una tabla con las frecuencias absoluta, relativa y un porcentaje, al final se añade una fila denominada **Total** para comprobar que hay 500 observaciones, que las proporciones suman 1 y que los porcentajes suman 100%. Las funciones **round(4)** y **round(2)** controlan únicamente la presentación de los decimales.

```python
tabla_frecuencias = pd.DataFrame({
    'Frecuencia absoluta': frecuencia_pd,
    'Frecuencia relativa': frecuencia_relativa_pd,
    'Porcentaje (%)': porcentaje_pd
})
tabla_frecuencias.loc['Total'] = [
    tabla_frecuencias['Frecuencia absoluta'].sum(),
    tabla_frecuencias['Frecuencia relativa'].sum(),
    tabla_frecuencias['Porcentaje (%)'].sum()
]
print(tabla_frecuencias.round({'Frecuencia relativa': 4, 'Porcentaje (%)': 2}))
```

### Diagrama de barras con Pandas

El comando **frecuencia_pd.plot(kind='bar')** crea un gráfico de barras directamente desde una Serie de Pandas, aquí se usa Matplotlib para poner un título, etiquetas y ajustar el diseño del gráfico.

```python
ax = frecuencia_pd.plot(kind='bar', figsize=(9, 5), color='steelblue')
ax.set_title('Diagrama de barras con Pandas')
ax.set_xlabel('Nivel de estudios')
ax.set_ylabel('Frecuencia absoluta')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Diagrama de sectores con Pandas

Igual que el comando anterior **frecuencia_pd.plot(kind='pie')** genera un gráfico circular, la función **autopct** muestra el porcentaje de cada categoría.

```python
ax = frecuencia_pd.plot(
    kind='pie', figsize=(7, 7),
    autopct='%1.2f%%',
    colors=['orange', 'green', 'blue', 'cyan', 'red']
)
ax.set_title('Diagrama de sectores con Pandas')
ax.set_ylabel('')
plt.tight_layout()
plt.show()
```

## Verificar que Pandas reproduce los resultados originales

Se realiza una comparación de la frecuencia obtenida con Pandas contra el conteo obtenido con NumPy anteriormente, el comando **assert** detiene la ejecución si los resultados no coinciden y si todo está correcto, se imprime un mensaje de confirmación.

```python
frecuencia_numpy = pd.Series(frecuencia, index=estudios).sort_index()
assert frecuencia_pd.equals(frecuencia_numpy), 'Las frecuencias no coinciden.'
assert int(frecuencia_pd.sum()) == 500, 'El total debe ser 500.'
print('Verificación correcta: Pandas y NumPy producen las mismas frecuencias.')
```
